```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

df = pd.read_csv('../data/netflix_catalogue.csv')
print(f"Loaded: {len(df)} titles")
print(df['type'].value_counts())
print(df.head())
```

```python
print("Genres:", df['genre'].value_counts().head(8))
print("\nCountries:", df['country'].value_counts().head(8))
print("\nRatings:", df['rating'].value_counts())
```

## Task 1 — Heatmap: content by rating and release decade

**What to build:** A heatmap showing the number of titles by **content rating** (y-axis) and **decade** (x-axis).

**Requirements:**
- Create a 'decade' column: `df['decade'] = (df['release_year'] // 10 * 10).astype(str) + 's'`
- Filter to TV-14, TV-MA, PG-13, R, PG only (most common ratings)
- Sequential colour scale (Blues)
- Values shown in cells (`text_auto=True`)
- Insight title about which rating dominates which decade


```python
# Task 1 — Heatmap: content by rating and release decade

# Create decade column
df['decade'] = (df['release_year'] // 10 * 10).astype(str) + 's'

# Filter to most common ratings
ratings_to_keep = ['TV-14', 'TV-MA', 'PG-13', 'R', 'PG']
df_filtered = df[df['rating'].isin(ratings_to_keep)]

# Aggregate and pivot
rating_decade = df_filtered.groupby(['rating', 'decade']).size().reset_index(name='count')
pivot = rating_decade.pivot(index='rating', columns='decade', values='count').fillna(0)

# Sort rows by total
pivot = pivot.loc[pivot.sum(axis=1).sort_values(ascending=False).index]

# Build heatmap
fig = px.imshow(
    pivot,
    color_continuous_scale='Blues',
    aspect='auto',
    text_auto=True,
    labels={'color': 'Titles'},
    height=450, width=900
)

fig.update_traces(
    textfont=dict(size=10, color='white'),
    hovertemplate='<b>%{y}</b> — %{x}<br>Titles: %{z}<extra></extra>'
)

fig.update_layout(
    title='TV-MA dominates modern Netflix — mature content surged from the 2010s onwards',
    font=dict(family='Arial', size=12),
    coloraxis_showscale=False,
    margin=dict(l=80, r=40, t=60, b=60),
    xaxis=dict(title='Decade'),
    yaxis=dict(title=''),
    paper_bgcolor='white'
)

fig.show()
```

## Task 2 — Waterfall: Movie vs TV Show additions by year

**What to build:** A waterfall chart showing how Netflix's **Movie library** grew year by year (2015-2022).

**Requirements:**
- Filter to Movies only
- Group by `added_year`, count titles per year
- Final bar should be the cumulative total
- Green bars (additions), blue total
- Annotation on the year with the largest single addition
- Insight title naming the growth story


```python
# Task 2 — Waterfall: Movie library growth 2015–2022

# Filter to Movies only, then group by added_year
movies = df[df['type'] == 'Movie']
adds = movies.groupby('added_year').size().reset_index(name='new_titles')
adds = adds[adds['added_year'].between(2015, 2022)].copy()

# Find year with largest single addition (for annotation)
peak_year = adds.loc[adds['new_titles'].idxmax()]

# Cumulative total for the final bar
cumulative = adds['new_titles'].sum()

# Build waterfall using Graph Objects
trace = go.Waterfall(
    x=adds['added_year'].astype(str).tolist() + ['Total 2015–2022'],
    y=adds['new_titles'].tolist() + [cumulative],
    measure=['relative'] * len(adds) + ['total'],
    connector=dict(line=dict(color='#AAAAAA', dash='dot')),
    increasing=dict(marker_color='#70AD47'),   # green for additions
    totals=dict(marker_color='#2E75B6'),       # blue for total
    texttemplate='%{y:,}',
    textposition='outside'
)

fig = go.Figure(data=[trace])

# Annotate the peak year
fig.add_annotation(
    # x=str(int(peak_year['added_year'])),
    x=4,
    y=peak_year['new_titles'] + 360,
    text=f"<b>Peak year<br>{int(peak_year['new_titles']):,} titles</b>",
    showarrow=True,
    arrowhead=2,
    arrowcolor='#E63946',
    font=dict(color='#E63946', size=11, family='Arial'),
    bgcolor='white',
    bordercolor='#E63946',
    borderwidth=1,
    borderpad=4
)

fig.update_layout(
    title='Netflix movie library grew more than 8x between 2015 and 2022 — with a sharp peak mid-decade',
    plot_bgcolor='white',
    paper_bgcolor='white',
    font=dict(family='Arial', size=12),
    yaxis=dict(gridcolor='#EEEEEE', title='Movies Added'),
    xaxis=dict(title='Year Added', showgrid=False),
    margin=dict(l=60, r=40, t=60, b=40),
    showlegend=False,
    height=550
)

fig.show()
```
